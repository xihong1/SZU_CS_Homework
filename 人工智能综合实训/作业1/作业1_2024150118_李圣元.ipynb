4. 本节中定义了形状(2,3,4)的张量X。len(X)的输出结果是什么?

```python
X=torch.arange(24).reshape(2,3,4)
len(X)
```

由此可得len(X)的输出结果是2，与最外层维度对齐

5. 对于任意形状的张量X,len(X)是否总是对应于X特定轴的长度?这个轴是什么?

```python
X=X.reshape(4,6)
len(X)
```

对于任意形状的张量X，len(X)的结果总是对应于X特定轴的长度，这个轴是X最外围的维度大小

6. 运行A/A.sum(axis=1)，看看会发生什么。请分析一下原因？

```python
A=torch.arange(20).reshape(4,5)
A/A.sum(axis=1)
```

运行A/A.sum(axis=1)指令会报错，提示我们RuntimeError: The size of tensor a (5) must match the size of tensor b (4) at non-singleton dimension 1，因为我们A.sum(axis=1)指令形成一个维度为4的tensor，而我们A的最后一个维度是4，都不是1，所以没有办法进行广播机制。我们可以在A.sum(axis=1)时加上指令keepdims=True来产生(5,1)的变量，保持轴数不变，这样子就可以进行除法了。

```python
A/A.sum(axis=1,keepdim=True)
```

7. 考虑一个具有形状(2,3,4)的张量，在轴0、1、2上的求和输出是什么形状?

```python
X=torch.arange(24).reshape(2,3,4)
print(X.sum(axis=0).shape)
print(X.sum(axis=1).shape)
print(X.sum(axis=2).shape)
```

在轴0、1、2上求和输出的形状分别是(3,4)(2,4)(2,3)，因为sum求和沿轴进行求和，保持在该轴上的形状

8. 为linalg.norm函数提供3个或更多轴的张量，并观察其输出。对于任意形状的张量这个函数计算得到什么?

```python
A = torch.arange(24).float().reshape(2,3,4)
print(torch.linalg.norm(A))
A=torch.arange(16).float().reshape(2,2,2,2)
print(torch.linalg.norm(A))
```

对于函数linalg.norm在不提供任何参数的时候，我们默认使用l2范数，对于任何轴的张量，函数都会先默认展成一维向量然后对于任何数字进行平方求和再开方，得到一个数字。
